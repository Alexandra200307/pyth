```python
import random

def func(somth):
    if somth == 'умножение':
        for i in range(1, 9):
                for j in range(1, 9):
                    print(f"{i*j:4}", end = "")
                print()
    if somth == 'деление':
            for i in range(1, 9):
                    for j in range(1, 9):
                        print(f"{i/j:4}", end = "")
                    print()
    if somth == 'сложение':
            for i in range(1, 9):
                    for j in range(1, 9):
                        print(f"{i+j:4}", end = "")
                    print()
    if somth == 'вычитание':
                for i in range(1, 9):
                        for j in range(1, 9):
                            print(f"{i-j:4}", end = "")
                        print()

func(input())
print()

def vect(N):
    a = []
    for i in range(N):
        x = random.uniform(0, 1)
        a.append(x)
    return a

print(vect(5))
print()

def matrix(m, n):
    return [[random.uniform(0, 1) for i in range(n)] for j in range(m)]


mat = matrix(3, 4)

for row in mat:
    print(row)

def mat_vec(A, v):
    m = len(A)       
    n = len(A[0]) 
    res = []

    for i in range(m):
        s = 0
        for j in range(n):
            s += A[i][j] * v[j]
        res.append(s)

    return res


A = [[1, 2, 3],
    [4, 5, 6]]

v = [10, 20, 30]

result = mat_vec(A, v)
print(result)
print()

def print_matrix(mat):
    for row in mat:
        print(row)

print_matrix(A)
print()

def print_vector(vec):
    for x in vec:
        print(x)

print_vector(v)
print()

def sum_diag(mat):
    s = 0
    for i in range(min(len(mat), len(mat[0]))):
        s += mat[i][i]
    return s

result = sum_diag(A)
print(result)
print()

def svertka(mat, kernel):
    kernel = [row[::-1] for row in kernel[::-1]]

    h, w = len(mat), len(mat[0])
    kh, kw = len(kernel), len(kernel[0])

    ph, pw = kh // 2, kw // 2

    result = [[0] * w for _ in range(h)]

    for i in range(h):
        for j in range(w):
            s = 0
            for ki in range(kh):
                for kj in range(kw):
                    y = i + ki - ph
                    x = j + kj - pw
                    if 0 <= y < h and 0 <= x < w:
                        s += mat[y][x] * kernel[ki][kj]
            result[i][j] = s

    return result

result = svertka(A,[[0.3, 0.3, 0.3], [0.3, 0.3, 0.3]])
print(result)
print()
```
